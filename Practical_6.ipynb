```python
def helpdesk():
    print("=== Help Desk Expert System ===")

    issue = input("Enter your issue: ").lower()

    if issue == "password":
        print("Solution: Reset your password and try again.")

    elif issue == "network":
        print("Solution: Check your internet connection.")

    elif issue == "printer":
        print("Solution: Check printer cable and restart printer.")

    elif issue == "software":
        print("Solution: Reinstall the software.")

    elif issue == "slow system":
        print("Solution: Close unwanted programs and restart system.")

    else:
        print("Solution not found. Contact technician.")


helpdesk()
```
