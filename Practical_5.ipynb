```python
def chatbot():
    print("Chatbot: Hello! Welcome to customer support.")
    print("Type 'bye' to exit.")

    while True:
        user = input("You: ").lower()

        if user == "hello" or user == "hi":
            print("Chatbot: Hello, how can I help you?")

        elif user == "price":
            print("Chatbot: Product price is Rs. 500.")

        elif user == "timing":
            print("Chatbot: We are open from 9 AM to 6 PM.")

        elif user == "location":
            print("Chatbot: We are located in Pune.")

        elif user == "contact":
            print("Chatbot: Contact us at 9876543210.")

        elif user == "bye":
            print("Chatbot: Thank you! Visit again.")
            break

        else:
            print("Chatbot: Sorry, I don't understand.")


chatbot()
```
