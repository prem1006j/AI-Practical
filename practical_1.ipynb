```python
graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C':['F'],
    'D':[],
    'E':[],
    'F':[]
}

visited = set()

def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        for neighbour in graph[node]:
            dfs(neighbour)

print("DFS Traversal:")
dfs('A')
```

```python
# BFS using queue

from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

visited = set()
queue = deque()

def bfs(start):
    visited.add(start)
    queue.append(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

print("BFS Traversal:")
bfs('A')
```
